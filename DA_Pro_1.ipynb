```python
'''
How to replace the NULL values ? BY mean or median

Generally, we should choose median over mean because mean is sensitive to outliers and that would skew the distribution of the data.
And median is robust to outliers. Don't fill the missing value with common median value, it's not the best praticse.
Instead we will imputte the missing values with the median review rating within each category.
'''
```

```python
# Imputing missing values in Review Rating column with the median rating of the product category

df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
# Renaming columns according to snake casing for better readability and documentation

df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
# create a new column age_group
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels)

# What does qcut do?
# It splits ages into 4 equal sized groups based on the data distribution and assigns the labels we defined.
```

```python
df[['age','age_group']].head(10)
```

```python
'''
Now we can see the original ages alongside the new age group column, hence easier to analyze patterns across diff age categories,
which can be really useful for marketing or understanding the customer behaviour
'''
```

```python
# create new column purchase_frequency_days

frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)

df.head()
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
df[['discount_applied','promo_code_used']].head(10)
```

```python
'''
So, at 1st glance, they kind of look like the same column, right?
Because if a promo code is used, then obviously a discount was applied.
But, sometimes businesses might apply discounts without promo codes like automatic seasonal sales or exclusive discounts for members
'''
```

```python
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
# Dropping promo code used column

df = df.drop('promo_code_used', axis=1)
```

```python
from sqlalchemy import create_engine

engine = create_engine(
    "postgresql+psycopg2://postgres:postgres@localhost:5432/customer_db"
)
```

```python
df.to_sql(
    "customer_shopping_behavior",
    engine,
    if_exists="replace",
    index=False
)

print("Data imported successfully!")
```

```python
query = """
SELECT *
FROM customer_shopping_behavior
LIMIT 10;
"""

result = pd.read_sql(query, engine)

result
```

```python
query = """
SELECT
    gender,
    COUNT(*) AS total_customers,
    ROUND(AVG(purchase_amount),2) AS avg_purchase
FROM customer_shopping_behavior
GROUP BY gender
ORDER BY avg_purchase DESC;
"""

pd.read_sql(query, engine)
```
